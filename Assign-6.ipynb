```python
# Q1a
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
corrected_temperature = temperature + 2

print("Original temperatures (C):", temperature)
print("Temperatures after adding 2 C:", corrected_temperature)
```

```python
# Q1b
fahrenheit = temperature * 9 / 5 + 32

print("Original temperatures (C):", temperature)
print("Temperatures in Fahrenheit:", fahrenheit)
```

```python
# Q1c
high_temperatures = temperature[temperature > 32]

print("Readings greater than 32 C:", high_temperatures)
```

```python
# Q1d
count = np.count_nonzero(temperature > 32)

print("Number of readings greater than 32 C:", count)
```

```python
# Q1e
print("Vectorization applies an operation to a whole array at once.")
print("NumPy does this work using optimized compiled code, avoiding the overhead of a Python loop.")
print("Boolean indexing creates a True/False mask and selects the values where the condition is True.")
print("For example, temperature[temperature > 32] selects all readings above 32 C in one expression.")
```

```python
# Q2a
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

total_steps = np.sum(steps)

print("Total steps recorded:", total_steps)
```

```python
# Q2b
mean_steps = np.mean(steps)

print("Mean number of steps:", mean_steps)
```

```python
# Q2c
maximum_steps = np.max(steps)
minimum_steps = np.min(steps)

print("Maximum steps:", maximum_steps)
print("Minimum steps:", minimum_steps)
```

```python
# Q2d
daily_steps = np.sum(steps, axis=0)

print("Total steps for days 1, 2 and 3:", daily_steps)
```

```python
# Q2e
user_steps = np.sum(steps, axis=1)

print("Total steps for users 1, 2, 3 and 4:", user_steps)
```

```python
# Q2f
maximum_index = np.argmax(steps)
user_index, day_index = np.unravel_index(maximum_index, steps.shape)
# position = np.argmax(steps)
# columns = steps.shape[1]

# user_index = position // columns
# day_index = position % columns
print("Flattened index of the maximum:", maximum_index)
print("User/day position (zero-based indices):", (user_index, day_index))
print("User number:", user_index + 1)
print("Day number:", day_index + 1)
print("Steps at this position:", steps[user_index, day_index])
```

```python
# Q4a
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])

print("Shape of X:", X.shape)
print("Number of dimensions of X:", X.ndim)
```

```python
# Q4b
X_transpose = X.T

print("Transpose of X:")
print(X_transpose)
print("The transpose swaps rows and columns, changing the shape from (4, 3) to (3, 4).")
print("Each row now contains one feature's values, and each column represents one user.")
```

```python
# Q4c
X_transpose_X = X.T @ X

print("X.T @ X:")
print(X_transpose_X)
```

```python
# Q4d
inverse = np.linalg.inv(X_transpose_X)

print("Inverse of X.T @ X:")
print(inverse)
print("X is not square, so it has no ordinary matrix inverse.")
print("X.T @ X is square and invertible for these data, so np.linalg.inv() can be used.")
```

```python
# Q4e
beta = inverse @ X.T @ y

print("OLS coefficients [sleep, activity, stress]:", beta)
print(f"Predicted score = ({beta[0]:.4f} * sleep hours) + ({beta[1]:.4f} * activity level) + ({beta[2]:.4f} * stress level)")
```

```python
# Q4f
print("Each coefficient is the change in the fitted score for a one-unit increase in its feature, keeping the other features fixed.")
print(f"One extra hour of sleep changes the predicted score by {beta[0]:.4f} points.")
print(f"One extra unit of activity changes the predicted score by {beta[1]:.4f} points.")
print(f"One extra unit of stress changes the predicted score by {beta[2]:.4f} points.")
print("These coefficients describe the fitted model; they do not prove that a feature causes the score to change.")
print("The supplied OLS equation fits these three features without an intercept term.")
```

```python
# Q4g
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("New user [sleep, activity, stress]:", new_user)
print(f"Predicted assistance score: {predicted_score:.4f}")
```
